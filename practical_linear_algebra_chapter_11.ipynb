# 1. What is a General Linear Model (GLM)?

A General Linear Model expresses the relationship between variables as:

\[
y = X\beta + \epsilon
\]

where:

- \(y\) : output vector
- \(X\) : design matrix
- \(\beta\) : parameter vector
- \(\epsilon\) : error term

This equation forms the foundation of many machine learning algorithms.


```
import numpy as np

X = np.array([
    [1,1],
    [1,2],
    [1,3],
    [1,4]
])

y = np.array([2,3,5,7])

print(X)
print(y)
```

The first column of ones represents the intercept term.


# 2. Overdetermined Systems

Often we have:

- More equations than unknowns.

In these situations:

\[
X\beta = y
\]

cannot be solved exactly.

Instead, we seek the best approximate solution.


# 3. Least Squares Principle

Least Squares minimizes the residual error:

\[
||X\beta-y||^2
\]

The goal is to find parameters that minimize the sum of squared errors.


```
beta = np.linalg.lstsq(
    X,
    y,
    rcond=None
)[0]

print(beta)
```

# 4. Geometric Interpretation

Least Squares projects the output vector onto the column space of X.

The projection minimizes the distance between:

- observed values
- predicted values


# 5. Normal Equation

The Least Squares solution can be computed using:

\[
\hat\beta=
(X^TX)^{-1}X^Ty
\]

This equation is called the Normal Equation.


```
beta = (
    np.linalg.inv(X.T @ X)
    @ X.T
    @ y
)

print(beta)
```

The result should match the solution obtained from `np.linalg.lstsq()`.


```
beta2 = np.linalg.lstsq(
    X,
    y,
    rcond=None
)[0]

print(np.allclose(beta,beta2))
```

# 6. Predictions

After estimating the coefficients:

\[
\hat y = X\hat\beta
\]


```
y_hat = X @ beta

print(y_hat)
```

# 7. Residual Errors

Residuals:

\[
e = y - \hat y
\]


```
residual = y - y_hat

print(residual)
print("SSE:", np.sum(residual**2))
```

# 8. Visualization of Least Squares


```
import matplotlib.pyplot as plt

x = X[:,1]

plt.scatter(x,y)
plt.plot(x,y_hat)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Least Squares Fit")
plt.show()
```

The line minimizes the sum of squared vertical distances from the observations.


# 9. Multiple Linear Regression

GLM can handle multiple features:

\[
y =
\beta_0+
\beta_1x_1+
\beta_2x_2+
...
\]

The matrix formulation remains exactly the same.


```
X = np.random.randn(100,3)

beta_true = np.array([2,-1,3])

y = X @ beta_true + np.random.randn(100)

beta_est = np.linalg.lstsq(
    X,
    y,
    rcond=None
)[0]

print(beta_est)
```
