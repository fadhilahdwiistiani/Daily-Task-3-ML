# 1. Orthogonal Vectors

Two vectors are orthogonal if their dot product equals zero.

Mathematically:

\[
u^Tv = 0
\]

Orthogonality means the vectors are perpendicular to each other.


```
import numpy as np

u = np.array([1,0])
v = np.array([0,1])

print("Dot product:", np.dot(u,v))
```

Because the dot product equals zero, the vectors are orthogonal.


# 2. Orthogonal Matrices

A matrix is orthogonal if:

\[
Q^TQ = QQ^T = I
\]

where:

- Q is an orthogonal matrix.
- I is the identity matrix.

An orthogonal matrix preserves lengths and angles.


```
theta = np.pi/4

Q = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta),  np.cos(theta)]
])

print(Q.T @ Q)
```

The result is approximately the identity matrix, verifying orthogonality.


# 3. Properties of Orthogonal Matrices

Orthogonal matrices satisfy:

1. Columns are orthonormal.
2. Rows are orthonormal.
3. Inverse equals transpose.

\[
Q^{-1}=Q^T
\]

This property makes computations more stable and efficient.


```
Q_inv = np.linalg.inv(Q)

print(Q_inv)
print(Q.T)
```

# 4. Why Orthogonal Matrices Matter

Orthogonal matrices are important because they:

- Preserve distances.
- Preserve angles.
- Avoid numerical instability.
- Simplify computations.


# 5. Rotation Matrices

Rotation matrices are examples of orthogonal matrices.

A rotation matrix:

\[
R=
\begin{bmatrix}
cos\theta & -sin\theta \\
sin\theta & cos\theta
\end{bmatrix}
\]

rotates vectors without changing their lengths.


```
import matplotlib.pyplot as plt

v = np.array([1,0])
new_v = Q @ v

plt.figure(figsize=(5,5))

plt.quiver(0,0,v[0],v[1],
           angles='xy',
           scale_units='xy',
           scale=1)

plt.quiver(0,0,new_v[0],new_v[1],
           angles='xy',
           scale_units='xy',
           scale=1)

plt.xlim(-1,1.5)
plt.ylim(-1,1.5)
plt.grid(True)
plt.show()
```

# 6. QR Decomposition

QR decomposition factors a matrix into:

\[
A = QR
\]

where:

- Q is an orthogonal matrix.
- R is an upper triangular matrix.


```
A = np.array([
    [1,2],
    [3,4],
    [5,6]
])

Q,R = np.linalg.qr(A)

print("Q:")
print(Q)

print("\nR:")
print(R)
```

# Verification of QR Decomposition


```
print(np.allclose(A, Q @ R))
```

The decomposition reconstructs the original matrix.


# 7. Geometric Interpretation

QR decomposition can be interpreted as:

- Q defines new orthogonal directions.
- R describes scaling along those directions.

This idea is fundamental in numerical linear algebra.


# 8. Gram-Schmidt Process

QR decomposition is closely related to the Gram-Schmidt orthogonalization procedure.

The algorithm:

1. Takes a set of vectors.
2. Produces orthogonal vectors.
3. Normalizes them into orthonormal vectors.
