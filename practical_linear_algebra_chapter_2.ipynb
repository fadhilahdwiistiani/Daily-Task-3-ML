# 1. What is a Vector?

A vector is an ordered collection of numbers arranged in a specific order.

Examples:

- Temperature measurements
- Pixel intensities in an image
- Features of a machine learning dataset
- Word embeddings in NLP

Mathematically:

\[
v = [1,2,3]
\]

This vector has three elements and therefore lives in a three-dimensional space.


```
import numpy as np

v = np.array([1,2,3])

print(v)
print("Dimension:", len(v))
```

## Code Explanation

`np.array()` creates a vector.

`len(v)` returns the number of elements or dimensions.


# 2. Column Vectors and Row Vectors

A vector can be represented as:

Row vector:

\[
[1,2,3]
\]

Column vector:

\[
\begin{bmatrix}
1 \\
2 \\
3
\end{bmatrix}
\]

Both represent the same information but are used differently in matrix operations.


```
row_vector = np.array([1,2,3])
column_vector = np.array([[1],[2],[3]])

print(row_vector.shape)
print(column_vector.shape)
```

# 3. Geometric Interpretation of Vectors

A vector can be interpreted as:

- A point in space
- A direction
- A displacement from the origin

Vectors are commonly visualized using arrows.


```
import matplotlib.pyplot as plt

v = np.array([3,2])

plt.figure(figsize=(5,5))
plt.quiver(0,0,v[0],v[1],angles='xy',scale_units='xy',scale=1)
plt.xlim(0,4)
plt.ylim(0,4)
plt.grid(True)
plt.show()
```

# 4. Vector Magnitude (Length)

The magnitude of a vector is:

\[
||v|| = \sqrt{v_1^2 + v_2^2 + ... + v_n^2}
\]

This is also called the Euclidean norm.


```
v = np.array([3,4])

magnitude = np.linalg.norm(v)

print("Magnitude:", magnitude)
```

For vector [3,4]:

\[
||v|| = \sqrt{3^2+4^2}=5
\]

This is the famous 3-4-5 right triangle.


# 5. Unit Vectors

A unit vector has magnitude equal to 1.

Formula:

\[
\hat v = \frac{v}{||v||}
\]


```
v = np.array([3,4])

unit_v = v / np.linalg.norm(v)

print(unit_v)
print(np.linalg.norm(unit_v))
```

# 6. Vector Addition

Two vectors can be added element by element.

\[
a+b=
[a_1+b_1,a_2+b_2]
\]


```
a = np.array([1,2])
b = np.array([3,1])

print(a+b)
```

# 7. Scalar Multiplication

Multiplying a vector by a scalar changes its length.

\[
2[1,2]=[2,4]
\]


```
v = np.array([1,2])

print(2*v)
print(-1*v)
```

# 8. Linear Combination

A linear combination is:

\[
c_1v_1+c_2v_2+...+c_nv_n
\]

This concept is fundamental in linear algebra and machine learning.


```
v1 = np.array([1,2])
v2 = np.array([3,1])

result = 2*v1 + 3*v2
print(result)
```
