Almost every dataset can be represented as a collection of vectors.


# 2. Data as Vectors

Suppose we have a dataset:

| Height | Weight | Age |
|---------|---------|-----|
| 170 | 65 | 20 |

This single observation can be represented as:

\[
x=[170,65,20]
\]

Every row in a dataset is essentially a vector.


```
import numpy as np

person = np.array([170,65,20])

print(person)
print("Dimensions:", len(person))
```

# 3. Vectors in Images

A grayscale image is simply a large vector of pixel values.

For example:

28 × 28 image:

\[
28 \times 28 = 784
\]

The image can be flattened into a vector with 784 elements.


```
image = np.random.randint(
    0,255,
    size=(28,28)
)

image_vector = image.flatten()

print(image_vector.shape)
```

# 5. Linear Independence

Vectors are linearly independent if none of them can be represented as a combination of the others.

Example:

\[
v_1=[1,0]
\]

\[
v_2=[0,1]
\]

These vectors are independent.


```
v1 = np.array([1,0])
v2 = np.array([0,1])

A = np.column_stack((v1,v2))

print(np.linalg.matrix_rank(A))
```

The matrix rank equals two, therefore the vectors are linearly independent.


# 6. Linear Dependence

Example:

\[
v_1=[1,2]
\]

\[
v_2=[2,4]
\]

The second vector is a multiple of the first one.


```
v1 = np.array([1,2])
v2 = np.array([2,4])

A = np.column_stack((v1,v2))

print(np.linalg.matrix_rank(A))
```

The rank becomes one because the vectors are dependent.


# 7. Basis Vectors

A basis is a set of independent vectors that span a vector space.

Standard basis in two dimensions:

\[
e_1=[1,0]
\]

\[
e_2=[0,1]
\]

Every vector in two-dimensional space can be expressed as a combination of these basis vectors.


```
e1 = np.array([1,0])
e2 = np.array([0,1])

v = 3*e1 + 2*e2

print(v)
```

# 8. Span of Vectors

The span of vectors consists of all possible linear combinations.

Example:

\[
c_1v_1+c_2v_2
\]

Changing the coefficients creates infinitely many vectors.


```
import matplotlib.pyplot as plt

x = np.linspace(-5,5,100)

plt.plot(x,2*x)
plt.grid(True)
plt.title("Span Example")
plt.show()
```

# 9. Dimensionality

Dimension is the number of independent directions in a space.

Examples:

- Line → 1 dimension
- Plane → 2 dimensions
- Space → 3 dimensions

Data science often deals with hundreds or thousands of dimensions.


# Curse of Dimensionality

As dimensions increase:

- Distances become less informative.
- Computation becomes expensive.
- Visualization becomes difficult.

Dimensionality reduction techniques such as PCA are therefore important.
