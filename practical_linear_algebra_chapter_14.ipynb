# 1. Introduction to Singular Value Decomposition

Singular Value Decomposition (SVD) is one of the most powerful tools in linear algebra.

Every matrix can be decomposed into three matrices:

\[
A = U\Sigma V^T
\]

where:

- U = left singular vectors
- Σ = singular values
- Vᵀ = right singular vectors


```
import numpy as np

A = np.array([
    [1,2],
    [3,4],
    [5,6]
])

U, S, VT = np.linalg.svd(A)

print("U:")
print(U)

print("\nSingular values:")
print(S)

print("\nV^T:")
print(VT)
```

# 2. Why SVD Matters

Unlike eigendecomposition, SVD exists for:

- Square matrices
- Rectangular matrices
- Rank-deficient matrices

This makes SVD one of the most versatile matrix decompositions.


# 3. Singular Values

Singular values describe the amount of stretching performed by the transformation.

The larger the singular value:

- the more important the corresponding direction.

Small singular values often represent noise.


```
print("Singular values:")
print(S)
```

# 4. Reconstruction of the Original Matrix

SVD can perfectly reconstruct the original matrix.


```
Sigma = np.zeros((A.shape[0], A.shape[1]))
Sigma[:len(S), :len(S)] = np.diag(S)

A_reconstructed = U @ Sigma @ VT

print(A_reconstructed)
print(np.allclose(A, A_reconstructed))
```

# 5. Geometric Interpretation

SVD can be interpreted as three successive operations:

1. Rotation using Vᵀ.
2. Scaling using Σ.
3. Rotation using U.

This interpretation makes SVD extremely useful in data analysis.


# 6. Matrix Rank

The rank of a matrix equals the number of non-zero singular values.


```
rank = np.linalg.matrix_rank(A)
print("Rank:", rank)
```

# 7. Low-Rank Approximation

One of the most important applications of SVD is low-rank approximation.

Instead of keeping all singular values, we keep only the most important ones.


```
k = 1

Uk = U[:, :k]
Sk = np.diag(S[:k])
VTk = VT[:k, :]

A_approx = Uk @ Sk @ VTk

print(A_approx)
```

Low-rank approximation compresses information while preserving most important patterns.


# 8. Dimensionality Reduction

SVD forms the mathematical foundation of:

- Principal Component Analysis (PCA)
- Latent Semantic Analysis (LSA)
- Recommendation Systems

The idea is to represent data using fewer dimensions.


```
import matplotlib.pyplot as plt

plt.plot(S, marker='o')
plt.title("Singular Values")
plt.xlabel("Index")
plt.ylabel("Magnitude")
plt.grid(True)
plt.show()
```

# 9. Image Compression Example

Images can be compressed using low-rank approximations.

By keeping only a few singular values:

- storage decreases,
- noise decreases,
- important structures remain.


```
img = np.random.rand(50,50)

U_img, S_img, VT_img = np.linalg.svd(img)

k = 10
img_approx = (
    U_img[:, :k]
    @ np.diag(S_img[:k])
    @ VT_img[:k, :]
)

plt.imshow(img_approx, cmap="gray")
plt.title("Compressed Image Approximation")
plt.show()
```
