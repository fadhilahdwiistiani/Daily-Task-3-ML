# 1. Why Are Matrices Important?

Matrices are one of the most important structures in modern computing.

Applications include:

- Machine Learning
- Deep Learning
- Computer Vision
- Graphics
- Recommendation Systems
- Signal Processing
- Scientific Computing

Almost every data science problem can be represented using matrices.


# 2. Datasets as Matrices

A dataset containing observations and features can naturally be represented as a matrix.

Example:

| Height | Weight | Age |
|--------|---------|-----|
|170|65|20|
|180|72|22|
|165|55|19|

becomes:

\[
X=
\begin{bmatrix}
170 & 65 & 20 \\
180 & 72 & 22 \\
165 & 55 & 19
\end{bmatrix}
\]


```
import numpy as np

X = np.array([
    [170,65,20],
    [180,72,22],
    [165,55,19]
])

print(X)
print("Shape:", X.shape)
```

Rows represent observations and columns represent features.

This representation is fundamental in machine learning.


# 3. Images as Matrices

A grayscale image is simply a matrix of pixel values.

Example:

\[
I=
\begin{bmatrix}
0 & 255 & 120 \\
30 & 90 & 220 \\
...
\end{bmatrix}
\]

Each number represents the brightness of a pixel.


```
import matplotlib.pyplot as plt

image = np.random.randint(
    0,255,
    size=(20,20)
)

plt.imshow(image, cmap='gray')
plt.title("Random Image Matrix")
plt.show()
```

# 4. Matrices in Computer Graphics

Matrices are used to perform:

- Rotation
- Scaling
- Reflection
- Translation

These operations are examples of linear transformations.


```
theta = np.pi/4

R = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta),  np.cos(theta)]
])

v = np.array([1,0])

print(R @ v)
```

The matrix rotates the vector by 45 degrees.


# 5. Covariance Matrices

Covariance matrices describe relationships between variables.

Formula:

\[
C=
\frac{1}{n-1}
X^TX
\]

Covariance matrices are central to:

- PCA
- Statistics
- Data analysis


```
X = np.random.randn(100,3)

cov = np.cov(X.T)

print(cov)
```

# 6. Matrices in Neural Networks

A neural network layer computes:

\[
y = Wx+b
\]

where:

- W is a weight matrix.
- x is the input vector.
- b is the bias vector.


```
W = np.random.randn(4,3)
x = np.random.randn(3)
b = np.random.randn(4)

y = W @ x + b

print(y)
```

Every modern deep learning model repeatedly performs matrix multiplication.


# 7. Recommendation Systems

Recommendation systems can be represented as matrices.

Example:

| User | Movie1 | Movie2 | Movie3 |
|------|---------|---------|---------|
| User1 | 5 | 3 | 1 |
| User2 | 4 | ? | 2 |
| User3 | 2 | 5 | 4 |

This is called a user-item matrix.


```
ratings = np.array([
    [5,3,1],
    [4,0,2],
    [2,5,4]
])

print(ratings)
```

Matrix factorization techniques such as SVD are commonly used to complete missing values.


# 8. Graphs and Networks

Adjacency matrices represent connections between nodes.

Example:

\[
A=
\begin{bmatrix}
0 & 1 & 1 \\
1 & 0 & 1 \\
1 & 1 & 0
\end{bmatrix}
\]


```
A = np.array([
    [0,1,1],
    [1,0,1],
    [1,1,0]
])

print(A)
```

Adjacency matrices are widely used in:

- Social networks
- Transportation systems
- Graph neural networks
