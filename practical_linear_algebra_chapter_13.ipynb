# 1. Introduction to Eigendecomposition

Eigendecomposition is one of the most important concepts in linear algebra.

Given a square matrix:

\[
A
\]

we seek vectors that preserve their directions after transformation.

These special vectors are called **eigenvectors**.


# 2. Eigenvalue Equation

The fundamental equation is:

genui{"math_block_widget_always_prefetch_v2":{"content":"A\mathbf{v}=\lambda\mathbf{v}"}}

where:

- \(A\) = transformation matrix
- \(\mathbf{v}\) = eigenvector
- \(\lambda\) = eigenvalue

The transformation changes only the magnitude of the vector while preserving its direction.


```
import numpy as np

A = np.array([
    [2,1],
    [1,2]
])

eigvals, eigvecs = np.linalg.eig(A)

print("Eigenvalues:")
print(eigvals)

print("\nEigenvectors:")
print(eigvecs)
```

# 3. Understanding Eigenvectors

An eigenvector does not rotate under the transformation.

Instead:

- it may stretch,
- shrink,
- or reverse direction.

The amount of scaling is determined by the eigenvalue.


# 4. Geometric Interpretation

Consider the matrix:

\[
A=
\begin{bmatrix}
2 & 1 \\
1 & 2
\end{bmatrix}
\]

Most vectors change direction after multiplication by A.

Only eigenvectors preserve their original directions.


```
import matplotlib.pyplot as plt

v = eigvecs[:,0]

new_v = A @ v

plt.figure(figsize=(5,5))

plt.quiver(0,0,v[0],v[1],
           angles='xy',
           scale_units='xy',
           scale=1)

plt.quiver(0,0,new_v[0],new_v[1],
           angles='xy',
           scale_units='xy',
           scale=1)

plt.grid(True)
plt.xlim(-2,2)
plt.ylim(-2,2)
plt.show()
```

The transformed vector lies on the same line as the original eigenvector.


# 5. Characteristic Equation

Eigenvalues are obtained from:

genui{"math_block_widget_always_prefetch_v2":{"content":"\det(A-\lambda I)=0"}}

This equation is called the characteristic equation.


# 6. Matrix Diagonalization

If a matrix has enough independent eigenvectors, then:

genui{"math_block_widget_always_prefetch_v2":{"content":"A=V\Lambda V^{-1}"}}

where:

- V contains eigenvectors.
- Λ is a diagonal matrix containing eigenvalues.


```
V = eigvecs
Lambda = np.diag(eigvals)

A_reconstructed = V @ Lambda @ np.linalg.inv(V)

print(A_reconstructed)
```

# Verification


```
print(np.allclose(A, A_reconstructed))
```

# 7. Why Eigendecomposition Matters

Eigendecomposition simplifies complicated matrix operations.

Applications include:

- Principal Component Analysis (PCA)
- Spectral Clustering
- Recommendation Systems
- Dynamical Systems
- Image Compression


# 8. Powers of Matrices

Using eigendecomposition:

\[
A^k=
V\Lambda^kV^{-1}
\]

Computing large matrix powers becomes much easier.


```
k = 5

A_power = np.linalg.matrix_power(A,k)

print(A_power)
```

# 10. Numerical Issues

Not all matrices:

- have real eigenvalues,
- have independent eigenvectors,
- are diagonalizable.

Numerical methods are often required for large matrices.
