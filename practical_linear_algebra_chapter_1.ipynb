# What is Linear Algebra?

Linear algebra studies vectors, matrices, and linear transformations.

It is the mathematical foundation of:

- Data Science
- Machine Learning
- Deep Learning
- Artificial Intelligence
- Computer Vision
- Scientific Computing

Modern linear algebra is computational and is best learned through coding and applications.


```
import numpy as np

A = np.array([[1,2],
              [3,4]])
x = np.array([5,6])

y = A @ x
print(y)
```

## Explanation

The code demonstrates:

\[
y = Ax
\]

where:

- A is a matrix,
- x is an input vector,
- y is the transformed vector.

This equation appears repeatedly throughout machine learning algorithms.

# Soft Proofs

The author proposes learning mathematics through experimentation in code.

Example:

\[
a \times b = b \times a
\]

Running the code multiple times builds intuition about commutativity.
