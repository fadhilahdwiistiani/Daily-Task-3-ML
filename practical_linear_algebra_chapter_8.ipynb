# 1. What is a Matrix Inverse?

The inverse of a matrix is analogous to division in ordinary arithmetic.

For a square matrix A:

\[
AA^{-1}=A^{-1}A=I
\]

where:

- \(A^{-1}\) is the inverse matrix.
- \(I\) is the identity matrix.


```
import numpy as np

A = np.array([
    [1,2],
    [3,4]
])

A_inv = np.linalg.inv(A)

print(A_inv)
```

# Verification

To verify that the matrix inverse is correct:

\[
AA^{-1}=I
\]


```
print(A @ A_inv)
print(A_inv @ A)
```

Because of numerical precision, the result may not be exactly equal to the identity matrix, but it should be extremely close.


# 2. Identity Matrix

The identity matrix acts like the number one in matrix algebra.

Example:

\[
I=
\begin{bmatrix}
1 & 0 \\
0 & 1
\end{bmatrix}
\]


```
I = np.eye(3)
print(I)
```

# 3. Conditions for Invertibility

A matrix has an inverse only if:

1. It is square.
2. Its determinant is non-zero.

Matrices that satisfy these conditions are called:

- Non-singular matrices
- Invertible matrices


```
A = np.array([
    [1,2],
    [3,4]
])

print(np.linalg.det(A))
```

# 4. Singular Matrices

A singular matrix does not have an inverse.

Example:

\[
A=
\begin{bmatrix}
1 & 2 \\
2 & 4
\end{bmatrix}
\]

The second row is a multiple of the first row.


```
A = np.array([
    [1,2],
    [2,4]
])

print(np.linalg.det(A))
```

The determinant becomes zero, therefore the matrix cannot be inverted.


```
try:
    print(np.linalg.inv(A))
except:
    print("Matrix is singular and cannot be inverted.")
```

# 5. Solving Linear Systems

Consider:

\[
Ax=b
\]

If A is invertible:

\[
x=A^{-1}b
\]


```
A = np.array([
    [2,1],
    [1,3]
])

b = np.array([8,13])

x = np.linalg.inv(A) @ b

print(x)
```

# Better Approach: Solve Directly

In practice, computing the inverse is often inefficient.

NumPy provides:

`np.linalg.solve()`


```
x = np.linalg.solve(A,b)
print(x)
```

This approach is:

- Faster
- More numerically stable
- Preferred in scientific computing


# 6. Geometric Interpretation

An inverse matrix reverses a transformation.

If:

\[
y=Ax
\]

then:

\[
x=A^{-1}y
\]

The inverse undoes scaling, rotation, and other transformations.


```
theta = np.pi/6

R = np.array([
    [np.cos(theta),-np.sin(theta)],
    [np.sin(theta), np.cos(theta)]
])

R_inv = np.linalg.inv(R)

print(R @ R_inv)
```

# 7. Matrix Inverse and Determinants

The determinant indicates:

- Whether an inverse exists.
- How much a transformation changes area or volume.

If:

\[
det(A)=0
\]

then:

\[
A^{-1}
\]

does not exist.


# 8. Pseudoinverse

Some matrices are not invertible.

In these situations we can use:

\[
A^{+}
\]

called the Moore-Penrose pseudoinverse.


```
A = np.array([
    [1,2],
    [2,4]
])

A_pinv = np.linalg.pinv(A)

print(A_pinv)
```

The pseudoinverse is widely used in:

- Linear regression
- Least squares
- Machine learning optimization
