# 1. Vector Transpose

The transpose operation converts a row vector into a column vector and vice versa.

Example:

Row vector:

\[
[1,2,3]
\]

Column vector:

\[
\begin{bmatrix}
1 \\
2 \\
3
\end{bmatrix}
\]

Transpose is denoted using:

\[
v^T
\]


```
import numpy as np

v = np.array([[1,2,3]])

print("Row vector:")
print(v)

print("\nTranspose:")
print(v.T)
```

# 2. Dot Product

The dot product is one of the most important operations in linear algebra.

Formula:

\[
a \cdot b = a_1b_1+a_2b_2+...+a_nb_n
\]

The result is always a scalar.


```
a = np.array([1,2,3])
b = np.array([4,5,6])

print(np.dot(a,b))
```

## Manual Computation

\[
1(4)+2(5)+3(6)
=4+10+18
=32
\]


# 3. Geometric Meaning of Dot Product

The dot product can also be written as:

\[
a \cdot b = ||a|| ||b|| cos(\theta)
\]

where:

- \(\theta\) is the angle between vectors.

This equation allows us to measure similarity between vectors.


```
a = np.array([1,2])
b = np.array([2,1])

theta = np.arccos(
    np.dot(a,b) /
    (np.linalg.norm(a)*np.linalg.norm(b))
)

print("Angle in radians:", theta)
print("Angle in degrees:", np.degrees(theta))
```

# 4. Orthogonal Vectors

Two vectors are orthogonal if:

\[
a \cdot b = 0
\]

Orthogonal vectors are perpendicular to each other.


```
a = np.array([1,0])
b = np.array([0,1])

print(np.dot(a,b))
```

The result is zero, therefore these vectors are orthogonal.


# 5. Vector Similarity

The dot product provides information about similarity.

- Positive value → similar direction.
- Zero → perpendicular.
- Negative value → opposite direction.

This idea is used heavily in:

- Recommendation systems
- Natural language processing
- Information retrieval


# 6. Cosine Similarity

Cosine similarity is defined as:

\[
cos(\theta)=
\frac{a \cdot b}
{||a|| ||b||}
\]


```
a = np.array([1,2,3])
b = np.array([2,4,6])

similarity = np.dot(a,b) / (
    np.linalg.norm(a) *
    np.linalg.norm(b)
)

print(similarity)
```

The result is close to one because the vectors point in the same direction.


# 7. Projection of One Vector onto Another

Projection measures how much of one vector lies in the direction of another vector.

Formula:

\[
proj_b a =
\frac{a \cdot b}
{||b||^2}
b
\]


```
a = np.array([3,4])
b = np.array([1,1])

projection = (
    np.dot(a,b) /
    np.dot(b,b)
) * b

print(projection)
```

# 8. Why Projection Matters

Applications:

- Signal processing
- Principal Component Analysis (PCA)
- Regression
- Computer graphics
- Neural networks


# 9. Euclidean Distance

Distance between vectors:

\[
d=
\sqrt{
\sum_i(a_i-b_i)^2
}
\]

Distance measures are fundamental in machine learning.


```
a = np.array([1,2,3])
b = np.array([2,4,6])

distance = np.linalg.norm(a-b)

print(distance)
```
