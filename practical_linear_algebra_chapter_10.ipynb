# 1. Systems of Linear Equations

Many machine learning problems can be expressed as systems of equations:

\[
Ax=b
\]

where:

- A is a matrix of coefficients.
- x is the unknown vector.
- b is the output vector.


```
import numpy as np

A = np.array([
    [2,1],
    [1,3]
])

b = np.array([8,13])

print(A)
print(b)
```

# 2. Elementary Row Operations

Three elementary operations:

1. Swap two rows.
2. Multiply a row by a nonzero scalar.
3. Add a multiple of one row to another row.

These operations preserve the solution of the system.


# 3. Augmented Matrix

A system of equations can be represented as:

\[
[A|b]
\]

Example:

\[
\begin{bmatrix}
2 & 1 & | & 8 \\
1 & 3 & | & 13
\end{bmatrix}
\]


```
aug = np.column_stack((A,b))
print(aug)
```

# 4. Row Echelon Form (REF)

A matrix is in REF if:

- Leading entries move to the right.
- Rows of zeros are at the bottom.
- Entries below pivots are zero.

REF simplifies solving linear systems.


# Example of REF

\[
\begin{bmatrix}
1 & 2 & 3 \\
0 & 1 & 4 \\
0 & 0 & 1
\end{bmatrix}
\]


# 5. Reduced Row Echelon Form (RREF)

A matrix is in RREF if:

- It satisfies REF.
- Every pivot equals one.
- Every pivot column contains zeros elsewhere.


# Why RREF Matters

RREF makes the solution of systems immediately visible.


```
from sympy import Matrix

A = Matrix([
    [2,1,8],
    [1,3,13]
])

print(A.rref())
```

# 6. Gaussian Elimination

Gaussian elimination uses row operations to transform matrices into REF.

Steps:

1. Create pivots.
2. Eliminate entries below pivots.
3. Perform back substitution.


# Example Solution

System:

\[
2x+y=8
\]

\[
x+3y=13
\]

Solution:

\[
x=2.2
\]

\[
y=3.6
\]


```
A = np.array([
    [2,1],
    [1,3]
])

b = np.array([8,13])

x = np.linalg.solve(A,b)

print(x)
```

# 7. LU Decomposition

LU decomposition factors a matrix into:

\[
A = LU
\]

where:

- L is a lower triangular matrix.
- U is an upper triangular matrix.


Triangular matrices are easier to solve than arbitrary matrices.


```
from scipy.linalg import lu

A = np.array([
    [2,1],
    [1,3]
])

P,L,U = lu(A)

print("L:")
print(L)

print("\nU:")
print(U)
```

# Verification

The original matrix can be reconstructed.


```
print(P @ L @ U)
```

# 8. Why LU Decomposition Matters

LU decomposition:

- Speeds up repeated computations.
- Reduces computational cost.
- Improves numerical efficiency.

It is widely used in scientific computing libraries.


# 9. Geometric Interpretation

Row reduction changes the representation of equations while preserving the solution.

LU decomposition rewrites the transformation into simpler triangular transformations.
