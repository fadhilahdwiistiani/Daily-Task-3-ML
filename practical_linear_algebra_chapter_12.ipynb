# 1. Why Least Squares Applications Matter

Least Squares is one of the most widely used optimization techniques in:

- Statistics
- Data Science
- Machine Learning

- Economics
- Computer Vision

Its objective is to minimize prediction errors.


# 2. Line Fitting Example

Suppose we have observations:

\[
y = mx+b
\]

Our goal is to estimate:

- slope \((m)\)
- intercept \((b)\)


```
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(1,11)
y = np.array([2,3,5,6,7,8,9,10,12,13])

X = np.column_stack((np.ones(len(x)), x))

beta = np.linalg.lstsq(X, y, rcond=None)[0]

print(beta)
```

The coefficients represent the intercept and slope of the best-fitting line.


```
y_hat = X @ beta

plt.scatter(x,y)
plt.plot(x,y_hat)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Least Squares Line Fitting")
plt.show()
```

# 3. Residual Errors

Residuals measure the difference between observations and predictions:

\[
e = y - \hat y
\]


```
residual = y - y_hat

print("Residuals:")
print(residual)

print("SSE:", np.sum(residual**2))
```

# 4. Polynomial Regression

Least Squares can fit nonlinear relationships by introducing polynomial features.

Example:

\[
y=
\beta_0+
\beta_1x+
\beta_2x^2
\]


```
X_poly = np.column_stack((
    np.ones(len(x)),
    x,
    x**2
))

beta_poly = np.linalg.lstsq(
    X_poly,
    y,
    rcond=None
)[0]

print(beta_poly)
```

```
y_poly = X_poly @ beta_poly

plt.scatter(x,y)
plt.plot(x,y_poly)
plt.title("Polynomial Regression")
plt.show()
```

# 5. Data Approximation

Least Squares is frequently used when:

- Measurements contain noise.
- Exact models do not exist.
- We need the best approximation.

The fitted model represents an optimal compromise.


# 6. Signal Processing Example

Signals are often contaminated by noise.

Least Squares can estimate the underlying trend.


```
t = np.linspace(0,10,200)

signal = np.sin(t)
noise = 0.2*np.random.randn(len(t))

y_noisy = signal + noise

plt.plot(t,y_noisy)
plt.title("Noisy Signal")
plt.show()
```

# 8. Model Evaluation

Two commonly used metrics:

### Sum of Squared Errors (SSE)

\[
SSE=
\sum e_i^2
\]

### Mean Squared Error (MSE)

\[
MSE=
\frac{1}{n}
\sum e_i^2
\]


```
mse = np.mean(residual**2)

print("MSE:", mse)
```

# 9. Overfitting and Underfitting

A model that is too simple:

- Underfits the data.

A model that is too complex:

- Overfits the data.

Least Squares models must balance complexity and generalization.


# 10. Geometric Interpretation

Least Squares computes an orthogonal projection of the target vector onto the column space of the design matrix.

This projection minimizes the Euclidean distance between predictions and observations.
