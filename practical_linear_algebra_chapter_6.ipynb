# 1. Matrix Addition

Two matrices can be added if they have the same dimensions.

Formula:

\[
A+B=
[a_{ij}+b_{ij}]
\]

Matrix addition is performed element by element.


```
import numpy as np

A = np.array([[1,2],
              [3,4]])

B = np.array([[5,6],
              [7,8]])

print(A+B)
```

# 2. Matrix Subtraction

Subtraction also requires matrices to have the same shape.

Formula:

\[
A-B=
[a_{ij}-b_{ij}]
\]


```
print(A-B)
```

# 3. Scalar Multiplication

A scalar multiplies every element of a matrix.

Example:

\[
2A=
\begin{bmatrix}
2 & 4 \\
6 & 8
\end{bmatrix}
\]


```
print(2*A)
print(-1*A)
```

# 4. Matrix Transpose

The transpose exchanges rows and columns.

Formula:

\[
A^T
\]

If:

\[
A=
\begin{bmatrix}
1 & 2 & 3 \\
4 & 5 & 6
\end{bmatrix}
\]

then:

\[
A^T=
\begin{bmatrix}
1 & 4 \\
2 & 5 \\
3 & 6
\end{bmatrix}
\]


```
A = np.array([[1,2,3],
              [4,5,6]])

print(A.T)
```

Transpose plays an important role in:

- Linear regression
- Covariance matrices
- PCA
- Neural networks


# 5. Element-wise Multiplication

NumPy allows element-wise multiplication.

This operation is different from matrix multiplication.


```
A = np.array([[1,2],
              [3,4]])

B = np.array([[5,6],
              [7,8]])

print(A*B)
```

# 6. Matrix Multiplication

Matrix multiplication is one of the most important operations in linear algebra.

Requirements:

If

\[
A_{m \times n}
\]

and

\[
B_{n \times p}
\]

then

\[
AB
\]

is defined and produces an:

\[
m \times p
\]

matrix.


```
A = np.array([[1,2],
              [3,4]])

B = np.array([[5,6],
              [7,8]])

print(A @ B)
```

# Manual Calculation

\[
AB=
\begin{bmatrix}
19 & 22 \\
43 & 50
\end{bmatrix}
\]

Each element is computed using row-column multiplication.


# 7. Properties of Matrix Multiplication

### Associative

\[
(AB)C=A(BC)
\]

### Distributive

\[
A(B+C)=AB+AC
\]

### Non-Commutative

Generally:

\[
AB \neq BA
\]


```
A = np.random.randn(3,3)
B = np.random.randn(3,3)

print(np.allclose(A@B, B@A))
```

The result is usually False because matrix multiplication is not commutative.


# 8. Matrix-Vector Multiplication

The equation

\[
y = Ax
\]

is fundamental in:

- Linear regression
- Neural networks
- Computer graphics
- Scientific computing


```
A = np.array([[1,2],
              [3,4]])

x = np.array([5,6])

print(A @ x)
```

# Visualization of Matrix Multiplication

Matrix multiplication can be interpreted as:

- Rotations
- Scaling
- Reflections
- Linear transformations


```
import matplotlib.pyplot as plt

v = np.array([1,1])

A = np.array([[2,0],
              [0,1]])

new_v = A @ v

plt.figure(figsize=(5,5))
plt.quiver(0,0,v[0],v[1],
           angles='xy',
           scale_units='xy',
           scale=1)

plt.quiver(0,0,new_v[0],new_v[1],
           angles='xy',
           scale_units='xy',
           scale=1)

plt.xlim(0,3)
plt.ylim(0,3)
plt.grid(True)
plt.show()
```
